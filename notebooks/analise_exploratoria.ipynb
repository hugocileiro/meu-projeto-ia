```python
# Análise Exploratória de Dados - Fase 1
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Bibliotecas importadas com sucesso!")
print(f"Pandas versão: {pd.__version__}")
print(f"NumPy versão: {np.__version__}")
```

```python
# Criando dados fictícios para análise
np.random.seed(42)

dados = pd.DataFrame({
    'idade': np.random.randint(18, 65, 200),
    'salario': np.random.randint(2000, 15000, 200),
    'tempo_empresa': np.random.randint(1, 30, 200),
    'setor': np.random.choice(['TI', 'RH', 'Financeiro', 'Marketing'], 200),
    'avaliacao': np.random.uniform(1, 5, 200).round(1)
})

print("📊 Dados criados!")
print(f"Shape: {dados.shape[0]} linhas e {dados.shape[1]} colunas")
print("\nPrimeiras 5 linhas:")
dados.head()
```

```python
# Estatísticas descritivas
print("📈 Estatísticas Descritivas:")
dados.describe()
```

```python
# Gráficos
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Histograma da idade
dados['idade'].hist(ax=axes[0, 0], bins=20, color='skyblue', edgecolor='black')
axes[0, 0].set_title('Distribuição de Idade')

# Salário por setor
dados.boxplot(column='salario', by='setor', ax=axes[0, 1])
axes[0, 1].set_title('Salário por Setor')
axes[0, 1].set_xlabel('')

# Correlação
sns.scatterplot(data=dados, x='idade', y='salario', hue='setor', ax=axes[1, 0])
axes[1, 0].set_title('Idade vs Salário')

# Média de avaliação por setor
dados.groupby('setor')['avaliacao'].mean().plot(kind='bar', ax=axes[1, 1], color='green')
axes[1, 1].set_title('Média de Avaliação por Setor')

plt.tight_layout()
plt.show()
```
